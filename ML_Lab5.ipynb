**Develop a program to implement k-Nearest Neighbour algorithm to classify the randomly generated 100 values of x in the range of [0,1]. Perform the following based on dataset generated. <br>
a. Label the first 50 points {x1, ……, x50} as follows: if (xi ≤ 0.5), then xi ∊ Class1, else xi ∊ Class2 <br>
b. Classify the remaining points: x51, ……, x100 using KNN. Perform this for k=1, 2, 3, 4, 5, 20, 30**

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Generate 100 random numbers between 0 and 1
data = np.random.rand(100)

# Label the first 50 values
labels = []
for x in data[:50]:
    if x <= 0.5:
        labels.append(1)
    else:
        labels.append(2)

# Prepare training data
train_data = data[:50].reshape(-1, 1)
train_labels = np.array(labels)

# Prepare test data
test_data = data[50:].reshape(-1, 1)
```

```python
# Different values of k
k_values = [1, 2, 3, 4, 5, 20, 30]

# Perform KNN Classification
for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(train_data, train_labels)

    predicted = knn.predict(test_data)

    print("\nResults for k =", k)
    print("Test value\tPredicted class")

    for value, label in zip(test_data.flatten(), predicted):
        print(f"{value:.3f}\t\tClass {label}")
```
